task-3

```python
df["horsepower"] = df["horsepower"].replace("?", np.nan)
```

```python
df["horsepower"] = pd.to_numeric(df["horsepower"])
```

```python
df = df.dropna()
```

task-4

```python
X = df["horsepower"]
y = df["mpg"]
```

task-5

```python
plt.scatter(X, y)
plt.xlabel("Horsepower")
plt.ylabel("MPG")
plt.title("Horsepower vs MPG")
plt.show()
```

task-6

```python
X_bias = np.c_[np.ones(len(X)), X]
```

task-7

```python
df["horsepower"].corr(df["mpg"])
```

task-8

```python
theta = np.linalg.inv(X_bias.T @ X_bias) @ X_bias.T @ y
```

```python
theta
```

task-9

```python
y_pred = X_bias @ theta
```

```python
y_pred[:10]
```

task10

```python
residuals = y - y_pred
```

task-11

```python
SSE = np.sum(residuals**2)
print(SSE)
```

```python
MSE = np.mean(residuals**2)
print(MSE)
```

```python
R2 = 1 - (SSE / np.sum((y - np.mean(y))**2))
print(R2)
```

task12

```python
plt.scatter(X, y)
plt.plot(X, y_pred, color="red")
plt.xlabel("Horsepower")
plt.ylabel("MPG")
plt.title("Simple Linear Regression")
plt.show()
```
